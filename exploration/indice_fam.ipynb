```python
# Standardise column names across all 3
rename_map = {
    "Libellé établissement": "name",
    "Adresse": "address",
    "Arrondissement": "arrondissement",
    "Code INSEE": "code_insee",
    "Année scolaire": "annee_scolaire",
    "Type établissement": "type",
    "geo_point_2d": "geo_point"
}

for df in [college_data, maternel_data, elementaire_data]:
    df.rename(columns=rename_map, inplace=True)

# Keep only needed columns
cols = ["name", "address", "arrondissement", "code_insee", 
        "annee_scolaire", "type", "geo_point"]

college_data   = college_data[cols]
maternel_data  = maternel_data[cols]
elementaire_data = elementaire_data[cols]

# Merge all 3
schools = pd.concat([college_data, maternel_data, elementaire_data], ignore_index=True)

# Keep most recent year per school
schools = schools.sort_values("annee_scolaire", ascending=False)
schools = schools.drop_duplicates(subset=["name", "address"], keep="first")

# Split geo_point_2d into lat and lng
schools[["lat", "lng"]] = schools["geo_point"].str.split(",", expand=True)
schools["lat"] = schools["lat"].astype(float)
schools["lng"] = schools["lng"].astype(float)

# Drop rows with missing coordinates
schools = schools.dropna(subset=["lat", "lng"])

# Drop original geo_point column
schools = schools.drop(columns=["geo_point"])

# Save
schools.to_csv("../data/silver/schools_merged.csv", index=False)

print(f"Total schools: {len(schools)}")
schools.head()
```

```python
import geopandas as gpd

iris = gpd.read_file("../data/silver/iris.geojson")
iris = iris[iris["dep"] == "75"].copy()

population = pd.read_csv("../data/silver/population_paris.csv")
schools = pd.read_csv("../data/silver/schools_merged.csv")
```

```python
# 3–4. School points (WGS84) + IRIS polygons buffered in meters (Lambert-93)
BUFFER_METERS = 500

schools_geo = gpd.GeoDataFrame(
    schools.reset_index(drop=True),
    geometry=gpd.points_from_xy(schools["lng"], schools["lat"]),
    crs="EPSG:4326",
)
schools_proj = schools_geo.to_crs(epsg=2154)

iris_proj = iris.to_crs(epsg=2154)
iris_buffer = iris_proj.copy()
iris_buffer["geometry"] = iris_buffer.geometry.buffer(BUFFER_METERS)
iris_buffer = iris_buffer[["code_iris", "geometry"]].rename(columns={"code_iris": "IRIS"})

# 5. Spatial join — count schools within each buffer
joined = gpd.sjoin(
    schools_proj,
    iris_buffer,
    how="inner",
    predicate="within",
)

school_counts = joined.groupby("IRIS").size().reset_index(name="school_count")
school_counts.head(5)
```
